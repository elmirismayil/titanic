```python
df["total_revenue"] = df["unit_price"]*df["quantity"]*(1-df["discount"]/100)
df.head(10)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['order_month'] = df['order_date'].dt.month
df.head(10)
```
